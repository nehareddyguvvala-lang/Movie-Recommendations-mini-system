```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
movies = [
    "RRR",
    "Baahubali",
    "Pushpa",
    "Jersey",
    "Inception",
    "Avatar",
    "Titanic",
    "3 Idiots",
    "Dangal",
    "Interstellar"
]

print("Movies:")
for i, movie in enumerate(movies, 1):
    print(i, movie)
```

```python
ratings = np.array([
    [5, 4, 5, 0, 0, 4, 0, 5, 0, 0],
    [4, 5, 0, 4, 5, 0, 0, 4, 5, 0],
    [0, 4, 5, 5, 0, 5, 4, 0, 0, 5],
    [5, 0, 4, 0, 5, 4, 0, 5, 4, 5],
    [0, 5, 0, 4, 4, 5, 5, 0, 4, 5]
])

rating_table = pd.DataFrame(
    ratings,
    index=["User 1", "User 2", "User 3", "User 4", "User 5"],
    columns=movies
)

rating_table
```

```python
print("Number of missing ratings:")

missing = np.sum(ratings == 0)

print(missing)
```

```python
data = ratings.astype(float).copy()

for i in range(data.shape[0]):
    user_ratings = data[i][data[i] > 0]
    average = np.mean(user_ratings)

    data[i][data[i] == 0] = average

print("Matrix after filling missing ratings:")
print(np.round(data, 2))
```

```python
U, S, VT = np.linalg.svd(data, full_matrices=False)

print("U Matrix:")
print(np.round(U, 2))

print("\nSingular Values:")
print(np.round(S, 2))

print("\nVT Matrix:")
print(np.round(VT, 2))
```

```python
k = 2

U_k = U[:, :k]
S_k = np.diag(S[:k])
VT_k = VT[:k, :]

predicted_ratings = U_k @ S_k @ VT_k

predicted_ratings = np.clip(
    predicted_ratings,
    1,
    5
)

print("Predicted Rating Matrix:")
print(np.round(predicted_ratings, 2))
```

```python
predicted_table = pd.DataFrame(
    np.round(predicted_ratings, 2),
    index=["User 1", "User 2", "User 3", "User 4", "User 5"],
    columns=movies
)

predicted_table
```

```python
user_number = int(input("Enter User Number (1-5): "))

user_index = user_number - 1

if user_index < 0 or user_index >= len(ratings):
    print("Invalid user number!")
else:
    print("Selected User:", user_number)
```

```python
print("Movies already rated by User", user_number)
print("--------------------------------------")

for i in range(len(movies)):
    if ratings[user_index][i] != 0:
        print(
            movies[i],
            "-> Rating:",
            ratings[user_index][i]
        )
```
